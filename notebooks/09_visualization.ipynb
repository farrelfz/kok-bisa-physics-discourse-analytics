# Stage 9: Research Visualization
This notebook generates publication-quality plots (Zipf's Law curve, Collocation Network, and UMAP/t-SNE semantic projection) and saves them in the `figures/` directory.

```python
import sys
import os
from pathlib import Path
import pandas as pd
import numpy as np

import sys, os
from pathlib import Path
cwd = Path(os.getcwd()).resolve()
base_dir = cwd if (cwd / 'config').exists() else (cwd.parent if (cwd.parent / 'config').exists() else cwd)
if str(base_dir) not in sys.path: sys.path.insert(0, str(base_dir))

from config import settings
from src.nlp.corpus_analyzer import fit_zipf_law, extract_collocations
from src.visualization.visualizer import plot_zipf, plot_embeddings_projection, plot_collocation_network
```

## 1. Load Analyzed Data & Embeddings

```python
df = pd.read_parquet(settings.PROCESSED_DATA_DIR / "comments_analyzed.parquet")
embeddings = np.load(settings.PROCESSED_DATA_DIR / "comments_embeddings.npy")
print(f"Loaded {len(df)} analyzed comments with embeddings {embeddings.shape}.")
```

## 2. Generate Zipf's Law Plot

```python
all_tokens = []
for t in df["tokens"].values:
    if t:
        all_tokens.extend(t.split(","))
        
ranks, frequencies, words = fit_zipf_law(all_tokens)
plot_zipf(ranks, frequencies, save_name="zipf_curve.png")
print("Zipf curve plotted.")
```

## 3. Generate Collocation Network

```python
collocs = extract_collocations(all_tokens, min_freq=2, top_n=20)
plot_collocation_network(collocs, save_name="collocation_network.png")
print("Collocation network plotted.")
```

## 4. Generate Semantic Projection Scatter Plot

```python
labels = df["discourse_act"].tolist()
plot_embeddings_projection(embeddings, labels, save_name="embeddings_projection.png")
print("Embedding projection plotted.")
```
