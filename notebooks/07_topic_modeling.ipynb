# Stage 7: Topic Modeling
This notebook runs Latent Dirichlet Allocation (LDA), Non-Negative Matrix Factorization (NMF), and Class-based TF-IDF cluster analysis (Embedding-based KMeans + c-TF-IDF Topic Model) to discover the underlying topics discussed in comments.

```python
import sys
import os
from pathlib import Path
import pandas as pd

import sys, os
from pathlib import Path
cwd = Path(os.getcwd()).resolve()
base_dir = cwd if (cwd / 'config').exists() else (cwd.parent if (cwd.parent / 'config').exists() else cwd)
if str(base_dir) not in sys.path: sys.path.insert(0, str(base_dir))

from config import settings
from src.nlp.topic_modeler import run_lda, run_nmf, run_mini_bertopic
from src.visualization.visualizer import plot_topics
```

## 1. Load Processed Comments

```python
df = pd.read_parquet(settings.PROCESSED_DATA_DIR / "comments_processed.parquet")
texts = df["cleaned_text"].tolist()
print(f"Loaded {len(texts)} comments for topic modeling.")
```

## 2. Run Latent Dirichlet Allocation (LDA)

```python
lda_model, lda_vect, lda_topics = run_lda(texts, n_topics=3)
print("LDA Topics:")
for topic in lda_topics:
    print(f"- Topic {topic['topic_id']}: {', '.join(topic['top_words'])}")
```

## 3. Run Non-Negative Matrix Factorization (NMF)

```python
nmf_model, nmf_vect, nmf_topics = run_nmf(texts, n_topics=3)
print("NMF Topics:")
for topic in nmf_topics:
    print(f"- Topic {topic['topic_id']}: {', '.join(topic['top_words'])}")
```

## 4. Run Embedding-based KMeans + c-TF-IDF Topic Model (Clustering + c-TF-IDF)

```python
mini_bertopic_topics = run_mini_bertopic(texts, n_topics=3)
print("Mini-BERTopic Topics:")
for topic in mini_bertopic_topics:
    print(f"- Topic {topic['topic_id']} (docs={topic['doc_count']}): {', '.join(topic['top_words'])}")
```

## 5. Export Topic Visualizations
This plots the NMF topics and saves them in the `figures/` directory.

```python
plot_topics(nmf_topics, save_name="topic_distribution.png")
print("Saved topic distribution plot to figures/topic_distribution.png.")
```
